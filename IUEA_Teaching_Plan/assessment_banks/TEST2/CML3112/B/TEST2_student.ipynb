## Understand the data
Each row is one synthetic observation of building cooling load. `record_id` identifies a record; it is not time. `equipment_id` identifies equipment and repeats across observations. `load_pct` is load in percent (valid range 0–100 inclusive). `ambient_c` is ambient temperature in degrees Celsius. `target` is building cooling load, measured in **kW**. Blank CSV values become missing values. These data are for teaching, not operational decisions.

A DataFrame is a table. The supplied helper converts it to a list of dictionaries: one dictionary per row, column names as keys, with missing values represented by `None`. Your functions accept these lists and return the stated structures. Keep function names, argument order and dictionary keys exactly as printed. Keep full numerical precision.

```python
import pandas as pd
import matplotlib.pyplot as plt

def records_from_table(table):
    objects = table.astype(object)
    return objects.where(table.notna(), None).to_dict("records")

raw_table = pd.read_csv("engineering_data.csv")
records = records_from_table(raw_table)
print(raw_table.head())
print("Rows:", len(raw_table))
```
