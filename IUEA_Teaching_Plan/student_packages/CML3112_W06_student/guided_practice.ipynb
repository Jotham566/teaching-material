## Cleaning order
*PowerPoint slide 12*

Keep the first record_id occurrence. Next, retain rows with a known target and load_pct between 0 and 100 inclusive.

Calculate the ambient_c median from those retained rows. fillna replaces only missing ambient readings.




## Operations for the next example
*PowerPoint slide 13*

Adding or replacing a column: table["name"] = values assigns that column. An array follows row order. A Series aligns by row label, so keep labels paired with the intended observations.

A separate working table: table.copy() creates a working copy. Assigning columns on this copy preserves the source table, so you can compare before and after.

Finding recorded cells: notna() returns True where a value is present. Use it to retain observations with a recorded target before modelling.




## Operations for the next example
*PowerPoint slide 14*

An inclusive range: series.between(0, 100) is True for values from 0 through 100, including both limits. The resulting mask helps select valid rows.

Middle recorded value: series.median() sorts known values conceptually and returns the middle one. With an even count, it averages the two middle values. Missing entries do not enter the calculation.

Replacing missing entries: series.fillna(value) returns a column whose missing entries contain value. Recorded values stay unchanged. Assign the returned column to retain the fill.




## Operations for the next example
*PowerPoint slide 15*

Rounded display: series.round(3) returns values rounded to three decimal places. Python round(number, 3) rounds one number. Keep unrounded calculations for returned assessment answers.

Keeping the first observation: table.drop_duplicates("record_id") returns rows with the first occurrence of each ID, preserving their order. It does not overwrite the source unless you assign the result.




## Cleaning the supplied observations
*PowerPoint slide 16*



Run this cell on the supplied CSV. Read the output below the cell.


```python
clean = raw.drop_duplicates("record_id").copy()
known = clean["target"].notna()
in_range = clean["load_pct"].between(0, 100)
clean = clean[known & in_range].copy()
fill = clean["ambient_c"].median()
clean["ambient_c"] = clean["ambient_c"].fillna(fill)
print(len(raw), len(clean), round(float(fill), 3))
print(clean.isna().sum())
```

## Checking numeric conversion

pd.to_numeric converts values to numbers. errors="coerce" turns invalid numeric text into a missing value.

Compare before and after conversion. A file with no invalid text should report zero conversion failures.




## Operations for the next example

Converting numeric text: pd.to_numeric(column, errors="coerce") converts numeric values or text. Unparseable text becomes missing. Compare original and converted values so existing gaps are not counted as conversion failures.




## Conversion check on the supplied ambient column
*PowerPoint slide 19*



Run this cell on the supplied CSV. Read the output below the cell.


```python
numbers = pd.to_numeric(raw["ambient_c"], errors="coerce")
failed = raw["ambient_c"].notna() & numbers.isna()
print(int(failed.sum()))
print(raw[failed].head())
```

## Summaries and grouped results
*PowerPoint slide 20*

mean, min and max summarise a numeric column. groupby collects records with the same equipment label before calculating.

The audit must inspect the supplied data for remaining defects, even when another function was meant to clean them.




## Operations for the next example
*PowerPoint slide 21*

Mean of recorded values: series.mean() adds known values and divides by their count. pandas skips missing values by default. The result has the same unit as the original column.

Several summaries at once: series.agg(["mean", "min", "max"]) applies those three named calculations and returns a labelled Series of results. min and max give the lowest and highest recorded values.

Grouping rows: table.groupby("label")["value"] groups the value column by matching labels. Apply mean() or sum() to calculate one result per group. Missing group labels are excluded by default.




## Summaries of the cleaned CSV
*PowerPoint slide 22*



Run this cell on the supplied CSV. Read the output below the cell.


```python
print(clean["target"].agg(["mean", "min", "max"]))
means = clean.groupby("equipment_id")["target"].mean()
print(means.round(3))
```

## A function called with the supplied records
*PowerPoint slide 31*



Output: {'E1': 54.9475, 'E2': 61.326, 'E3': 53.46933333333333, 'E4': 66.854, 'E5': 48.10133333333333, 'E6': 51.54}


```python
def mean_load_by_equipment(records):
    table = pd.DataFrame(records)
    result = table.groupby("equipment_id")["load_pct"].mean()
    return result.to_dict()
records = raw.to_dict("records")
print(mean_load_by_equipment(records))
```
