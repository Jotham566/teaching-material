## Worked example (1/2)
*PowerPoint slide 10*



Read the code together. Predict the result before running.


```python
from statistics import median
raw = [20, None, 22, 24, 80]
known = []
for value in raw:
    if value is not None:
        known.append(value)
fill = median(known)
cleaned = []
for value in raw:
    if value is None:
        cleaned.append(fill)
    else:
        cleaned.append(value)
print(fill)
print(raw)
print(cleaned)
```

## Copies and duplicate checks
*PowerPoint slide 15*



dict(row) makes a shallow copy. A set keeps unique IDs.


```python
row = {"id": "A", "reading": 4}
copy = dict(row)
copy["reading"] = 5
seen = set()
seen.add(row["id"])
print(row["reading"])
print("A" in seen)
```

## Copying records and recognising repeats
*PowerPoint slide 16*

set() creates a collection of already-seen identifiers. in checks membership. continue skips the rest of that iteration. dict(row) copies a record so later edits do not change the original.

Different record IDs from the same equipment remain legitimate observations.

## Copying records and recognising repeats in Python (1/2)
*PowerPoint slide 18*



Read the code together. Predict the result before running.


```python
raw_records = [{"id": 1, "equipment": "A", "value": 10},
               {"id": 1, "equipment": "A", "value": 10},
               {"id": 2, "equipment": "A", "value": 12}]
seen = set()
kept = []
for row in raw_records:
    if row["id"] in seen:
        continue
    seen.add(row["id"])
    kept.append(dict(row))
print(len(raw_records), len(kept))
```

## Summaries returned as a dictionary

## Summaries returned as a dictionary in Python
*PowerPoint slide 21*



Read the code together. Predict the result before running.


```python
def describe(values):
    count = len(values)
    mean = sum(values) / count
    return {"count": count, "mean": mean}

print(describe([10, 12]))
```
