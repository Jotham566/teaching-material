## Task 1. Load and inspect the file
Read `orders.csv` into a DataFrame called `orders` and print its shape. Then store the number of missing values in each column in `missing` and print it.

```python
import pandas as pd

orders = pd.read_csv('orders.csv')
print(orders.shape)
missing = orders.isna().sum()
print(missing)
```

```python
assert orders.shape == (30229, 7), 'Check the file name and the shape'
assert missing['customer_id'] == 9171, 'missing should count the blanks in every column. customer_id has 9171'
print('Task 1 correct')
```

## Task 2. Jinja orders
Branch B04 is Jinja. Keep its orders in a table called `jinja` and count them.

```python
jinja = orders[orders.branch_id == 'B04']
print(len(jinja))
```

```python
assert len(jinja) == 3562, 'Use a double equals sign and the text B04 in quotes'
print('Task 2 correct')
```

## Task 3. The ten largest orders
Sort `orders` by `order_value_ugx`, largest first, and keep the top ten in `top10`.

```python
top10 = orders.sort_values('order_value_ugx', ascending=False).head(10)
top10[['order_id', 'branch_id', 'order_value_ugx']]
```

```python
assert top10.iloc[0].order_id == 'O014303', 'The largest order should be O014303'
print('Task 3 correct')
```

## Task 4. Jinja and mobile money
How many Jinja orders were paid by `Mobile money`? Store the count in `n_mm`.

```python
n_mm = len(orders[(orders.branch_id == 'B04') & (orders.payment_method == 'Mobile money')])
print(n_mm)
```

```python
assert n_mm == 1695, 'Use & and put each test in round brackets'
print('Task 4 correct')
```
