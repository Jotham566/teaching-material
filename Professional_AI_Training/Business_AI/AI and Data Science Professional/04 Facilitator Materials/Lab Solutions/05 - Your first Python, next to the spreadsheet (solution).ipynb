## Task 2. A list of prices
Make a list called `prices` with 3200, 8400, 8700, 7000, 6300. Store the sum in `total` and the average in `average`.

```python
prices = [3200, 8400, 8700, 7000, 6300]
total = sum(prices)
average = sum(prices) / len(prices)
print(total, average)
```

```python
assert total == 33600, 'Use sum(prices)'
assert average == 6720, 'Divide the sum by len(prices)'
print('Task 2 correct')
```

## Task 3. A dictionary price list
Make `price_list` with the codes P001 to P005 as labels and the same five prices as values. Look up P003 into `p003`. Add P006 with the price 3600.

```python
price_list = {'P001': 3200, 'P002': 8400, 'P003': 8700, 'P004': 7000, 'P005': 6300}
p003 = price_list['P003']
price_list['P006'] = 3600
print(p003, len(price_list))
```

```python
assert p003 == 8700, 'Look up the label P003 in square brackets'
assert len(price_list) == 6, 'Add P006 so the list has six entries'
print('Task 3 correct')
```
