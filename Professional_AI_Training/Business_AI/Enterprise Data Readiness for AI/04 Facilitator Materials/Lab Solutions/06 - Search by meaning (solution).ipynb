## Task 1. Keyword ranks for the first ten questions
Build the searcher. For each of the first ten questions in `eval_questions.csv`, find the place of the right document in the top ten keyword results (0 if it is missing) and keep it in the dictionary `bm25_rank`.

```python
import pandas as pd
import kitara_search as ks

corpus = ks.load_corpus('.')
chunks = ks.build_chunks(corpus)
search = ks.Searcher(chunks)
tests = pd.read_csv('eval_questions.csv').head(10)

def rank_of_gold(question, gold, method):
    ids = list(search.search(question, method, k=10).doc_id)
    return ids.index(gold) + 1 if gold in ids else 0

bm25_rank = {r.q_id: rank_of_gold(r.question, r.gold_doc_id, 'bm25') for r in tests.itertuples()}
print(bm25_rank)
```

```python
assert bm25_rank['Q01'] == 7 and bm25_rank['Q03'] == 1, 'Q01 should be 7 and Q03 should be 1 with keyword search'
assert sum(1 for v in bm25_rank.values() if v == 0) == 0, 'All ten right documents appear in the top ten'
print('Task 1 correct')
```

## Task 2. Where does meaning search help?
Do the same with the stand-in meaning search into `meaning_rank`. Then list in `rescued` the questions where keyword search puts the right document outside the top five (or misses it) and meaning search puts it inside the top five.

```python
meaning_rank = {r.q_id: rank_of_gold(r.question, r.gold_doc_id, 'meaning') for r in tests.itertuples()}
rescued = [q for q in bm25_rank if (bm25_rank[q] == 0 or bm25_rank[q] > 5) and 0 < meaning_rank[q] <= 5]
print(meaning_rank)
print(rescued)
```

```python
assert meaning_rank['Q01'] == 1 and meaning_rank['Q02'] == 1, 'Meaning search should put Q01 and Q02 first'
assert rescued == ['Q01', 'Q02'], f'Expected Q01 and Q02 to be rescued, got {rescued}'
print('Task 2 correct')
```

## Task 3. Fusion by hand for Q01
The right chunk for Q01 is `D007-01`. Find its place in the keyword list and in the meaning list (use `k=50`), add the two scores 1 / (60 + place), round to four decimals into `fused`, and find the place of the right document in the hybrid search into `hybrid_rank`.

```python
q = tests.question.iloc[0]
def chunk_place(method):
    ids = list(search.search(q, method, k=50).chunk_id)
    return ids.index('D007-01') + 1
place_kw, place_mn = chunk_place('bm25'), chunk_place('meaning')
fused = round(1 / (60 + place_kw) + 1 / (60 + place_mn), 4)
hybrid_rank = rank_of_gold(q, 'D007', 'hybrid')
print(place_kw, place_mn, fused, hybrid_rank)
```

```python
assert (place_kw, place_mn) == (7, 2), f'Expected places 7 and 2, got {place_kw} and {place_mn}'
assert fused == 0.0311, f'fused should be 0.0311, got {fused}'
assert hybrid_rank == 3, 'The right document is third in the hybrid search'
print('Task 3 correct')
```

## Task 4. The Swahili questions
For the six questions with `language` equal to `sw` in `eval_questions.csv`, find the place of the right document in the top ten of the hybrid search (0 if missing) into `sw_hybrid`. Put the questions found in the top five into `found`.

```python
all_tests = pd.read_csv('eval_questions.csv')
sw = all_tests[all_tests.language == 'sw']
sw_hybrid = {r.q_id: rank_of_gold(r.question, r.gold_doc_id, 'hybrid') for r in sw.itertuples()}
found = sorted(q for q, r in sw_hybrid.items() if 0 < r <= 5)
print(sw_hybrid)
print(found)
```

```python
assert len(sw_hybrid) == 6, 'There are six Swahili questions'
assert found == ['Q49', 'Q50', 'Q51'], f'Only the three questions on Swahili circulars are found, got {found}'
print('Task 4 correct')
```

## Task 5. A real embedding model (needs a download, run it last)
This cell downloads a small English model the first time and needs the internet. Embed all 139 chunks and check the shape. The rest of the notebook does not depend on it. Read the model card for the licence before you use it in your organisation.

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
vectors = model.encode(list(chunks.text), normalize_embeddings=True)
query_vector = model.encode(q, normalize_embeddings=True)
nearest = list((vectors @ query_vector).argsort()[::-1][:3])
print(vectors.shape, list(chunks.chunk_id[nearest]))
```

```python
assert vectors.shape == (139, 384), f'Expected (139, 384), got {vectors.shape}'
print('Task 5 correct')
```
