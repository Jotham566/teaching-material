## Task 1. Rank the items
Read `roadmap_items.csv` into `items`. Add a column `priority` equal to value divided by effort_days times 10, rounded to 2 places. Store the ids of the three highest priorities, best first, in `top3`.

```python
import pandas as pd

items = pd.read_csv('roadmap_items.csv')
items['priority'] = (items.value / items.effort_days * 10).round(2)
top3 = list(items.sort_values('priority', ascending=False).id.head(3))
print(top3)
```

```python
assert len(items) == 16, 'There are 16 work items'
assert top3 == ['W01', 'W11', 'W16'], 'Sort by priority, highest first, and take the first three ids'
print('Task 1 correct')
```

## Task 2. The chains that set the end date
Build `deps`, a dictionary from each id to the list of ids it waits for. Write a function `chains(i)` that returns every chain of items ending at `i`. Store the length in months of the longest chain ending at W16 in `longest`, and the number of chains with that length in `n_longest`.

```python
deps = {r.id: str(r.depends_on).split(';') if r.depends_on == r.depends_on else []
        for r in items.itertuples()}
months = items.set_index('id').duration_months

def chains(i):
    if not deps[i]:
        return [[i]]
    return [c + [i] for d in deps[i] for c in chains(d)]

lengths = [int(sum(months[x] for x in p)) for p in chains('W16')]
longest = max(lengths)
n_longest = lengths.count(longest)
print(longest, n_longest)
```

```python
assert longest == 12, 'The longest chain to W16 takes 12 months'
assert n_longest == 3, 'Three different chains reach 12 months'
print('Task 2 correct')
```

## Task 3. Effort per month
Spread each item's effort evenly over the months it runs. Store the effort of each month 1 to 12 in a Series `load` and the busiest month in `busiest`.

```python
load = pd.Series(0.0, index=range(1, 13))
for r in items.itertuples():
    for m in range(r.start_month, r.start_month + r.duration_months):
        load[m] += r.effort_days / r.duration_months
busiest = load.idxmax()
print(load.round(1).to_dict())
print(busiest)
```

```python
assert round(load.sum()) == 276, 'The monthly loads add up to 276 person-days'
assert busiest == 4 and round(load[4], 1) == 53.3, 'Month 4 needs 53.3 person-days'
print('Task 3 correct')
```

## Task 4. Draw the roadmap
Draw a horizontal bar chart with one bar per item, starting at `start_month` and running for `duration_months`. Store it in `ax`.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(items.id, items.duration_months, left=items.start_month - 1)
ax.invert_yaxis()
ax.set_xlim(0, 12)
ax.axvline(6, linestyle='--')
plt.tight_layout()
plt.show()
```

```python
assert len(ax.patches) == 16, 'One bar for each of the 16 items'
assert ax.get_xlim()[1] == 12, 'Set the x axis from 0 to 12 months'
print('Task 4 correct')
```

## Task 5. The projected rescored baseline
Read `scorecard_after.csv`. Store the mean of `projected_score_month_12` in `target_mean` and the dimension with the largest `change_from_baseline` in `biggest`. Remember that the result is a target, not a measurement.

```python
sc = pd.read_csv('scorecard_after.csv')
target_mean = round(sc.projected_score_month_12.mean(), 2)
biggest = sc.loc[sc.change_from_baseline.idxmax(), 'dimension']
print(target_mean, biggest)
```

```python
assert target_mean == 3.3, 'The mean of the five projected scores is 3.3'
assert biggest == 'People and governance', 'People and governance gains most, 1.5'
print('Task 5 correct')
```
