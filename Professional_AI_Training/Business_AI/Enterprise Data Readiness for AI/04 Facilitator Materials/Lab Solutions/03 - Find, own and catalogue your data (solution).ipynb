## Task 1. Scan the folder tree
Use `os.walk` to count the files under `docs`. Store the total in `n_files`, the count by file ending in `by_type` and the count by folder name in `by_folder`.

```python
import os
from collections import Counter

by_type = Counter()
by_folder = Counter()
for folder, _, files in os.walk('docs'):
    for name in files:
        by_type[os.path.splitext(name)[1]] += 1
        by_folder[os.path.basename(folder)] += 1
n_files = sum(by_type.values())
print(n_files, dict(by_type))
print(by_folder.most_common(3))
```

```python
assert n_files == 121, 'Docs holds 121 files. Add up by_type'
assert by_type['.pdf'] == 105, 'There are 105 PDF files. Use os.path.splitext(name)[1]'
assert by_folder['policies'] == 37, 'The policies folder holds 37 files'
print('Task 1 correct')
```

## Task 2. Find the gaps in the inventory
Read `sources.csv`. Store the ids of the sources with no owner in `no_owner`, and the ids with no sensitivity class in `no_class`, each as a list.

```python
import pandas as pd

src = pd.read_csv('sources.csv')
no_owner = src[src.owner.isna()].source_id.tolist()
no_class = src[src.sensitivity.isna()].source_id.tolist()
print(no_owner)
print(no_class)
```

```python
assert no_owner == ['S08', 'S09', 'S12', 'S13'], 'Four sources have no owner. Filter with src.owner.isna()'
assert no_class == ['S06', 'S08', 'S09', 'S12', 'S13'], 'Five sources have no class'
print('Task 2 correct')
```

## Task 3. Count the blanks per source
Count the blank cells in `owner`, `steward`, `sensitivity` and `personal_data` for each row, as the workbook's Missing fields column does. Store the total in `total_blanks` and the number of rows with at least one blank in `rows_with_gap`.

```python
key = src[['owner', 'steward', 'sensitivity', 'personal_data']]
per_row = key.isna().sum(axis=1)
total_blanks = int(per_row.sum())
rows_with_gap = int((per_row > 0).sum())
print(total_blanks, rows_with_gap)
```

```python
assert total_blanks == 19, 'The four columns hold 19 blanks in total'
assert rows_with_gap == 7, 'Seven rows have at least one blank'
print('Task 3 correct')
```

## Task 4. Catch classes that break the standard
The Data Classification Standard puts salary data and bulk national ID numbers in confidential. Find the sources whose `notes` mention `salary` or `national ID` and whose class is not confidential. Store their ids in `wrong_class`.

```python
hits = src[src.notes.str.contains('salary|national ID', case=False, na=False)]
wrong_class = hits[hits.sensitivity != 'confidential'].source_id.tolist()
print(wrong_class)
```

```python
assert wrong_class == ['S03', 'S10'], 'Two sources are recorded below confidential: S03 and S10'
print('Task 4 correct')
```

## Task 5. Follow the lineage
Read `lineage_edges.csv`. Starting at `claims_system.claims`, follow the edges and store every asset downstream in `downstream`. Store the number of steps that a person does by hand (the step text contains `by hand` or `pasted`) in `hand_steps`.

```python
edges = pd.read_csv('lineage_edges.csv')
todo, downstream = ['claims_system.claims'], []
while todo:
    a = todo.pop()
    for b in edges[edges.from_asset == a].to_asset:
        if b not in downstream:
            downstream.append(b)
            todo.append(b)
hand_steps = int(edges.step.str.contains('by hand|pasted').sum())
print(len(downstream), hand_steps)
```

```python
assert len(downstream) == 7, 'Seven assets sit downstream of claims_system.claims'
assert 'board.quarterly_pack' in downstream, 'The board pack is reached through the dashboard'
assert hand_steps == 3, 'Three edges are done by hand or pasted'
print('Task 5 correct')
```
