## Step 7: Why not the raw dot product?

```python
factor = 20
E2 = E.copy()
E2[5] = factor * E2[5]                  # a longer picnic vector
dots = E2 @ q
sims2 = dots / (np.linalg.norm(E2, axis=1) * np.linalg.norm(q))
```
