## Step 5: Draw the next word: slices and random numbers

The next-word list of Session 13 (illustrative). Each word owns a slice of 0 to 1 as wide as its probability; the slice edges are the running totals. A random number u picks the first word whose running total reaches u. The helper `draw` below does that for a list of u values.

**PREDICT first.** Which word owns the biggest slice? In 8 draws, about how many times do you expect tea?

Write your guess here before you run the next cell:

```python
words = ["tea", "coffee", "water", "milk", "soup"]
z = np.array([3.0, 2.6, 1.9, 1.2, 0.4])
e = np.exp(z - z.max())
p1 = e / e.sum()                       # stable softmax (Session 12)
u = np.random.default_rng(19).random(8)  # eight fixed random numbers

def draw(cum, us):
    """For each u, the first word whose running total reaches u."""
    return [words[i] for i in np.searchsorted(cum, us)]

print(np.round(p1, 4), np.round(u, 3))
```
