## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| project question | One question a table can answer with numbers. | Which item sells most, asked of a till roll. |
| scope | How much the project covers, and what it leaves out. | The fence round a garden. |
| data dictionary | A list of every column with its meaning and unit. | The key to a map. |
| limitation | What the data or method cannot tell us. | The part of the map that is blank. |
| licence | The rules that say how you may use a dataset. | The terms on a borrowed book. |
| disclosure | A plain statement of any help or AI tool you used. | Writing your sources at the end of an essay. |
