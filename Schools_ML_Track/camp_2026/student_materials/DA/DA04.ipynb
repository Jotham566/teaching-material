## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| comparison | A question that gives True or False, such as units >= 5. | Asking whether a pupil is tall enough to ride. |
| Boolean | A value that is either True or False. | A light that is on or off. |
| if | Run the indented lines only when the condition is True. | Enter only if you have a ticket. |
| else | Run these lines when the condition is False. | Otherwise, go to the queue. |
| indentation | Four spaces at the start of a line show which lines belong to the if. | The inside of a bracket in a sentence. |
| and, or | and needs both conditions True. or needs at least one. | Entry needs a ticket and an ID. A discount needs a student card or a coupon. |


## Block A. Compare and decide

A comparison asks a question. == asks whether two values are equal. >= means at least. The word if runs its indented lines only when the answer is True. A colon ends the if line and four spaces mark the lines that belong to it.

**Predict.** Predict what prints for a budget of 4999, then for 5000, then for 5001. Write your guess in the cell below before you run anything.

My prediction:


```python
budget = 5000
print(budget >= 5000)
if budget >= 5000:
    print("Buy")
else:
    print("Save")
```
