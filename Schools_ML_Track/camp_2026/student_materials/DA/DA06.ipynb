```python
def revenue(units, price):
    return units * price

print(revenue(3, 2000))
assert revenue(3, 2000) == 6000
assert revenue(0, 2000) == 0
```

## Block B. Edge cases and a guard

A share is a part divided by a whole. Dividing by zero is an edge case. A function can guard against it with an if before it divides.

**Predict.** Predict the result of share(12, 20) and of share(5, 0). Write your guess in the cell below before you run anything.

My prediction:


```python
def share(part, whole):
    if whole == 0:
        return 0
    return part / whole

print(share(12, 20))
print(share(5, 0))
assert share(12, 20) == 0.6
assert share(5, 0) == 0
```

**Run it, then compare.** The cell prints 0.6 and then 0. When the whole is zero the function returns before it divides, so no error appears.

**Investigate.**

1. Why does the function check the whole before it divides?
2. Is returning 0 for share(5, 0) always the right answer?
