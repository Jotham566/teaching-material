## Block A. Bar and line charts

plt.bar draws one bar for each category. A bar chart starts its value axis at zero, because the length of a bar stands for the amount. plt.plot joins points in order, which suits dates.

**Predict.** Predict which of the three bars will be tallest. Write your guess in the cell below before you run anything.

My prediction:


```python
import pandas as pd
import matplotlib.pyplot as plt
shop = pd.read_csv("camp_shop.csv")
shop["revenue_ugx"] = shop["units"] * shop["unit_price_ugx"]
by_item = shop.groupby("item")["revenue_ugx"].sum()
plt.bar(by_item.index, by_item.values)
plt.ylabel("Revenue (UGX)")
plt.title("Revenue by item, practice shop")
plt.ylim(bottom=0)
plt.show()
by_date = shop.groupby("date")["revenue_ugx"].sum()
plt.plot(by_date.index, by_date.values)
plt.ylabel("Revenue (UGX)")
plt.xticks(rotation=90)
plt.show()
```

**Run it, then compare.** The first chart has three bars. Book is tallest at 297000, then Juice and Pen. The second chart is a line of 20 daily revenues, with the highest point on 2026-12-20. Every axis says what it measures and the bar chart starts at zero.

1. Why does the bar chart start at zero?
2. Why use a line for dates and bars for items?

## Block B. Histogram and scatter plot

A histogram shows how the values are spread. A scatter plot puts two numbers on the two axes. These two charts use the Sunrise Shops tables.

**Predict.** Predict the shape of the histogram of sale values: balanced, or bunched at the low end with a long tail? Write your guess in the cell below before you run anything.

My prediction:


```python
sunrise = pd.read_csv("sunrise_sales.csv")
sunrise["line_value"] = sunrise["units"] * sunrise["unit_price_ugx"]
plt.hist(sunrise["line_value"], bins=30)
plt.xlabel("Value of one sale (UGX)")
plt.ylabel("Number of sales")
plt.show()
daily = pd.read_csv("sunrise_daily_summary.csv")
plt.scatter(daily["customers_counted"], daily["revenue_ugx"])
plt.xlabel("Customers counted in a day")
plt.ylabel("Revenue that day (UGX)")
plt.show()
```

**Run it, then compare.** The histogram is bunched at the low end with a long tail to the right, because most sales are small and a few are very large. The scatter plot has 180 dots, one for each shop and day. More customers usually mean more revenue, but the dots are scattered, so the link is not exact.
