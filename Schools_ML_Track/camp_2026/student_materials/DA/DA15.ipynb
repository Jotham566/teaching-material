## Understand the idea

SUM adds values; COUNT(*) counts records. GROUP BY creates a result per category. AS gives a result column a readable name. An expression such as units * unit_price_ugx is calculated for each record before SUM combines it.



## Worked task
SELECT item, SUM(units * unit_price_ugx) AS revenue_ugx FROM sales GROUP BY item ORDER BY revenue_ugx DESC, item;

Expected result: One row per item, ordered by total revenue. Totals match the pandas summary.
