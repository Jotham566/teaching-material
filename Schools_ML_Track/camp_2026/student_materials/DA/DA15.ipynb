## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| correlation | A number from -1 to 1 for how closely two columns move together. | Taller pupils usually have longer feet. |
| strength | How close the correlation is to 1 or -1. Near 0 means weak. | How tightly the dots hug a line. |
| heatmap | A table of numbers coloured by size. | A map shaded by rainfall. |
| hue | A colour that shows which group each dot belongs to. | A different coloured pen for each shop. |
| cause | One thing makes another happen. A correlation alone does not show it. | Rain makes the ground wet. Wet ground does not make rain. |


## Block A. A scatter plot and one correlation

sns.scatterplot puts one column on each axis. hue= colours the dots by group. .corr() calculates the correlation between two columns.

**Predict.** Predict whether the correlation between customers and revenue is close to 0, close to 1 or close to -1. Write your guess in the cell below before you run anything.

My prediction:


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
daily = pd.read_csv("sunrise_daily_summary.csv")
sns.scatterplot(data=daily, x="customers_counted", y="revenue_ugx", hue="branch")
plt.show()
r = daily["customers_counted"].corr(daily["revenue_ugx"])
print(round(r, 3))
```

**Run it, then compare.** The dots rise from left to right, so more customers usually mean more revenue. The correlation is 0.623: positive and moderately strong. The dots are scattered, so a customer count does not fix the revenue.

## Block B. A heatmap of every pair, and the rain question

A heatmap of .corr() shows every pair at once. annot=True writes the numbers in the squares. Rain is in the data as rain_mm. Does it matter?

**Predict.** Predict whether rain_mm is strongly related to customers. Write your guess in the cell below before you run anything.

My prediction:


```python
cols = ["customers_counted", "rain_mm", "sale_lines", "revenue_ugx"]
table = daily[cols].corr()
sns.heatmap(table, annot=True, vmin=-1, vmax=1)
plt.show()
print(round(table.loc["rain_mm", "customers_counted"], 3))
print(round(table.loc["sale_lines", "revenue_ugx"], 3))
```

**Run it, then compare.** Rain and customers have a correlation of -0.013, which is essentially zero. Sale lines and revenue have 0.565. In this data the weather barely matters. A weak correlation is a finding too.
