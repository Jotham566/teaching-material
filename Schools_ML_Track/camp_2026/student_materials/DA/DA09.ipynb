## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| group | Gather rows with the same label before you calculate. | Sorting the sales into piles by item. |
| sort | Put rows in order from smallest to largest or the other way. | Lining up pupils by height. |
| mean | The total shared out equally: add the values and divide by how many. | Sharing a bag of mandazi equally. |
| median | The middle value once the values are in order. | The pupil standing in the middle of a line sorted by height. |
| proportion | A part divided by the whole. Times 100 it becomes a percentage. | 12 pupils out of 20 passed. |
| denominator | The whole that a proportion is measured against. | The 20 in 12 out of 20. |


## Block A. Group, total and sort

groupby gathers the rows with the same item. A calculation such as sum then works inside each group. sort_values puts the answer in order.

**Predict.** Predict which item sold the most units and which earned the most money. Write your guess in the cell below before you run anything.

My prediction:


```python
import pandas as pd
shop = pd.read_csv("camp_shop.csv")
shop["revenue_ugx"] = shop["units"] * shop["unit_price_ugx"]
units_by_item = shop.groupby("item")["units"].sum()
revenue_by_item = shop.groupby("item")["revenue_ugx"].sum().sort_values(ascending=False)
print(units_by_item)
print(revenue_by_item)
```

**Run it, then compare.** Pen sold the most units, 103. Book earned the most money, 297000 shillings, then Juice 150000 and Pen 103000. The three revenues add to 550000. The item that sells the most units is not the item that earns the most.

## Block B. Mean, median and a proportion

Now meet the bigger Sunrise Shops table: 1513 sales from three shops. The mean can be pulled up by a few huge sales. The median is not. A proportion needs its whole named: 36 large orders out of 1513 sales.

**Predict.** Predict whether the mean sale value is above or below the median. Write your guess in the cell below before you run anything.

My prediction:


```python
sunrise = pd.read_csv("sunrise_sales.csv")
sunrise["line_value"] = sunrise["units"] * sunrise["unit_price_ugx"]
mean_value = sunrise["line_value"].mean()
median_value = sunrise["line_value"].median()
bulk = (sunrise["units"] >= 10).sum()
share = bulk / len(sunrise)
print(round(mean_value), median_value)
print(bulk, len(sunrise), round(share * 100, 1))
```

**Run it, then compare.** The mean sale value is about 4214 shillings and the median is 2400. A few bulk purchases pull the mean up. 36 of the 1513 sales have 10 units or more, which is 2.4 percent. The 1513 is the denominator.
