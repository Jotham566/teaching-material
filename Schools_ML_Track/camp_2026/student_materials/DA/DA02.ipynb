| Word | What it means | Think of it like this |
|---|---|---|
| value | One piece of information, such as 2000 or Juice. | One entry on a receipt. |

| float | A number with a decimal part. | A price of 2500.5 shillings. |
| string | Text inside quotation marks. | A name written on a label. |
| variable | A name that points to a value. | A label on a box of stock. |
| type | The kind of value: integer, float or string. | Knowing whether a box holds pens or juice. |
