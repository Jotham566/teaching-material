A revenue column multiplies the units and price in each row: sales["revenue_ugx"] = sales["units"] * sales["unit_price_ugx"]. Selecting an existing column reads it; assigning a new column name creates it. Revenue is money from sales, not profit: costs are absent. ## Worked example
Predict the output, then run the following cell.

```python
import pandas as pd
import matplotlib.pyplot as plt
sales = pd.read_csv("camp_shop.csv")
sales["revenue_ugx"] = sales["units"] * sales["unit_price_ugx"]
by_item = sales.groupby("item")["revenue_ugx"].sum()
print(sales.shape[0])
print(sales["revenue_ugx"].sum())
print(by_item)
assert by_item.sum() == sales["revenue_ugx"].sum()
plt.bar(by_item.index, by_item.values)
plt.ylabel("Revenue (UGX)")
plt.title("Revenue by product — synthetic practice shop")
plt.show()
```

Expected result: 60 rows; product totals add to the overall revenue. Read the exact totals from the printed output and compare with the source rows.
