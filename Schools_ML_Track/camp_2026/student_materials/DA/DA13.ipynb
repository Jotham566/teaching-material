## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| NumPy | The library of fast number tools that pandas is built on. We load it as np. | The calculator inside the register. |
| array | A row of numbers that NumPy can calculate with all at once. | A strip of numbers on a ruler. |
| element-wise | Doing the same calculation to every number in an array at once. | Adding 100 shillings to every price on the board. |
| percentile | The value below which a given share of the numbers fall. | The mark that 75 out of 100 pupils did not beat. |
| standard deviation | A number for how far values typically sit from their mean. | How far pupils usually stand from the middle of the line. |
| outlier | A value far from the rest. | One huge order in a week of small sales. |


## Block A. Arrays, summaries and percentiles

Each shop counted its customers every day for 60 days: 180 counts in all. .to_numpy() turns a column into an array. np.mean and np.median give the usual summaries and np.percentile gives the value below which a share of the counts fall.

**Predict.** Predict whether the mean or the median of the daily counts is bigger. Write your guess in the cell below before you run anything.

My prediction:


```python
import numpy as np
import pandas as pd
daily = pd.read_csv("sunrise_daily_summary.csv")
customers = daily["customers_counted"].to_numpy()
print(customers.shape)
print(round(np.mean(customers), 1), np.median(customers))
print(np.percentile(customers, [25, 75]))
print(customers.max() - customers.min())
print(np.mean(customers) == daily["customers_counted"].mean())
```

**Run it, then compare.** The array has shape (180,). The mean is 143.0 and the median 131.5, so the mean is bigger. The 25th and 75th percentiles are 86.75 and 203.75: a quarter of the days had fewer than 87 customers and a quarter had more than 204. The range is 332. The last line prints True, because pandas and NumPy give the same mean.
