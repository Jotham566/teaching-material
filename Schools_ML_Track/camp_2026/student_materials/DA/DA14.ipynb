## Block A. A histogram of sale values

sns.histplot draws a histogram from a DataFrame. Name the DataFrame in data= and the column in x=. A map turns branch ids into names.

**Predict.** Predict whether the histogram is balanced or skewed to the right. Write your guess in the cell below before you run anything.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sunrise = pd.read_csv("sunrise_sales.csv")
sunrise["line_value"] = sunrise["units"] * sunrise["unit_price_ugx"]
sunrise["branch"] = sunrise["branch_id"].map({1: "Central Market", 2: "Campus Gate", 3: "Roadside"})
sns.histplot(data=sunrise, x="line_value", bins=30)
plt.xlabel("Value of one sale (UGX)")
plt.show()
print(round(sunrise["line_value"].skew(), 1))
```

**Run it, then compare.** The histogram is bunched at the low end with a long tail to the right. The skew is 7.3. A positive skew above 1 means a long right tail. Most sales are small and a few are large.

## Block B. A box plot for each shop

sns.boxplot draws one box for each group. The line inside the box is the median. The box runs from the lower quartile to the upper quartile. Dots beyond the whiskers are outliers.

**Predict.** Predict which shop has the highest median sale. Write your guess in the cell below before you run anything.

My prediction:


```python
sns.boxplot(data=sunrise, x="branch", y="line_value")
plt.ylabel("Value of one sale (UGX)")
plt.show()
medians = sunrise.groupby("branch")["line_value"].median()
print(medians)
```

**Run it, then compare.** The medians are 2500 at Central Market, 2000 at Campus Gate and 2500 at Roadside. The boxes look alike and every shop has dots well above its whiskers: bulk sales happen in all three. The shops differ more in how many sales they make than in how big a typical sale is.
