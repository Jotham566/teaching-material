## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| select | Name a column in square brackets to look only at it. | Reading only the price strip of the register. |
| filter | Keep only the rows where a comparison is True. | Picking out the sales of 5 units or more. |
| mask | A column of True and False values, one for each row, that a filter uses. | A ticked list beside the register. |
| new column | A column worked out from other columns and added to the table. | Multiplying units by price on every line. |
| revenue | Money received from sales: units times price. It is not profit. | The takings before paying for stock. |
| sum | All the values added together. | Adding up the day's takings. |
