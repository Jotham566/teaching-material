## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| raw data | The file exactly as it arrived. You never overwrite it. | The original receipt book. |
| missing value | A cell with nothing recorded. It is not the same as zero. | A blank in the register. |
| duplicate | A row that repeats another row exactly. | The same sale written twice. |
| data type | How a column is stored: whole number, decimal or text. | Whether a box holds numbers or words. |
| copy | A separate table to work on, so the raw table stays untouched. | A photocopy of the receipt book. |
| cleaning log | A record of each problem found and what was done about it. | The tailor's note of every change to a uniform. |


## Block A. Missing values and duplicates in a small table

Keep the raw table unchanged and work on a copy. isna marks missing cells True. Adding the True values counts them. drop_duplicates removes rows that repeat another row exactly.

**Predict.** Predict how many missing prices there are and how many rows stay after the duplicate is removed. Write your guess in the cell below before you run anything.

My prediction:


```python
import pandas as pd
raw = pd.DataFrame({"id": [1, 2, 2, 3], "price": [1000, None, None, 3000]})
clean = raw.drop_duplicates().copy()
print(raw["price"].isna().sum())
print(len(clean))
print(clean["price"].mean())
```

**Run it, then compare.** The raw table has 2 missing prices. After the duplicate row is removed, 3 rows stay and one price is still missing. The mean of the two known prices is 2000. The missing price is unknown and is not free.

## Block B. The raw Sunrise file

sunrise_sales_raw.csv is the Sunrise sales as an untidy export. Count the problems. to_csv saves a table to a file. index=False stops pandas adding an extra column of row numbers.

**Predict.** Predict how many duplicate rows the file holds. The clean version has 1513 sales. Write your guess in the cell below before you run anything.

My prediction:


```python
raw_sales = pd.read_csv("sunrise_sales_raw.csv")
print(raw_sales.shape)
print(raw_sales.duplicated().sum())
deduped = raw_sales.drop_duplicates()
print(deduped.shape)
print(deduped["units"].isna().sum())
log = pd.DataFrame({"problem": ["duplicate rows", "blank units"], "rows": [raw_sales.duplicated().sum(), deduped["units"].isna().sum()]})
log.to_csv("cleaning_log.csv", index=False)
print(pd.read_csv("cleaning_log.csv"))
```

**Run it, then compare.** The raw file has 1538 rows. 25 are exact duplicates, so removing them leaves 1513, the same as the clean file. 40 rows have blank units. The log table is saved and read back with two lines: 25 duplicate rows and 40 blank-unit rows.

1. Why does the cleaned row count of 1513 match the clean file?
