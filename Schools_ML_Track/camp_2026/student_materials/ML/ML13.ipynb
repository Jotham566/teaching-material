## Block A. The baseline guess

A baseline uses no features at all. It guesses the average of the training targets every time.

**Predict.** Training prices are 2000, 4000 and 6000. Predict the baseline guess and its error when the real price is 5000. Write your guess in the cell below before you run anything.

My prediction:


```python
train_prices = [2000, 4000, 6000]
baseline = sum(train_prices) / len(train_prices)
actual = 5000
error = abs(actual - baseline)
print(baseline, error)
```

**Run it, then compare.** The cell prints 4000.0 and 1000.0. The training mean 4000.0 is the baseline guess. abs removes the minus sign, so the error is the distance between the guess and the real value, 1000 shillings. A model is worth using only if its errors on unseen examples are smaller than the baseline's.

## Block B. Open the session 1 model

The model from session 1 had one feature and one target. Print what it learned.

**Predict.** Name the feature and the target in this code before you run it. Write your guess in the cell below before you run anything.

My prediction:


```python
import pandas as pd
from sklearn.linear_model import LinearRegression
homes = pd.read_csv("synthetic_homes.csv")
X = homes[["area_m2"]]
y = homes["rent_ugx"]
model = LinearRegression().fit(X, y)
slope = model.coef_[0]
intercept = model.intercept_
print(round(slope), round(intercept))
```

**Run it, then compare.** The cell prints 2996 and 159436. The feature is area_m2, kept in X. The target is rent_ugx, kept in y. fit looked at all 120 homes and found a line: rent is about 159436 shillings plus 2996 shillings for each square metre. That model was scored on nothing at all. Judging it on the same homes it learned from would flatter it, so next session we keep some homes back.
