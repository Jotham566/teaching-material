## Block B. A loop that adds

Run the loop and follow how total grows after each pass.

**Predict.** Predict the four numbers printed. Write your guess in the cell below before you run anything.

My prediction:


```python
prices = [1000, 2000, 3000]
total = 0
for price in prices:
    print(price * 2)
    total = total + price
print(total)
```
