## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| bar chart | Bars compare a number across categories. | Stacks of coins, one stack per item. |
| histogram | Bars show how many values fall in each range. | Sorting pupils into groups by height. |
| box plot | A box shows the middle half of the values, with a line at the median. | A summary of a class's marks in one picture. |
| axis label | Words that name what an axis measures, with its unit. | The heading on a column in a register. |
| title | One line saying what the chart shows. | The heading of a notice. |
| finding | A claim a chart supports, with a number. | Book earns the most money: UGX 297000. |


## Block A. A bar chart of revenue

Chart the revenue of each item from the practice shop. Label the axis with its unit.

**Predict.** Predict which bar will be tallest and roughly how tall. Write your guess in the cell below before you run anything.

My prediction:


```python
import matplotlib.pyplot as plt
import pandas as pd
shop = pd.read_csv("camp_shop.csv")
shop["revenue_ugx"] = shop["units"] * shop["unit_price_ugx"]
totals = shop.groupby("item")["revenue_ugx"].sum()
plt.bar(totals.index, totals.values)
plt.ylabel("Revenue (UGX)")
plt.title("Revenue by item in the practice shop")
plt.ylim(bottom=0)
plt.show()
```

**Run it, then compare.** Three bars appear. Book is tallest at 297000, then Juice at 150000 and Pen at 103000. plt is the chart tool from matplotlib. ylabel and title add words, and ylim(bottom=0) starts the axis at zero so bar heights are fair. One finding is: Book earns the most, UGX 297000 of 550000. One limitation is: the data is invented and has no costs.

## Block B. A histogram and a box plot

seaborn draws statistical charts and builds on matplotlib. Draw how units vary and how they differ by item.

**Predict.** Predict which item has the lowest median units. Write your guess in the cell below before you run anything.

My prediction:


```python
import seaborn as sns
sns.histplot(data=shop, x="units", discrete=True)
plt.title("Units per sale")
plt.show()
sns.boxplot(data=shop, x="item", y="units")
plt.title("Units per sale for each item")
plt.show()
medians = shop.groupby("item")["units"].median()
```

**Run it, then compare.** The histogram counts sales for each number of units and shows how they spread from 1 to 8. The box plot shows one box per item with a line at its median. Juice has the lowest median, 3.5 units, while Book has 5.0 and Pen 5.5. seaborn plots read straight from the table, which saves writing the grouping yourself.
