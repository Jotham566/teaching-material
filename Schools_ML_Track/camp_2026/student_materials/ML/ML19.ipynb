```python
import pandas as pd
fares = pd.read_csv("synthetic_fares.csv")
print(fares.shape)
corr = fares[["distance_km", "rush_hour", "raining", "tip_ugx", "fare_ugx"]].corr()["fare_ugx"].round(2)
print(corr)
```

**Run it, then compare.** The table has 150 rows and 6 columns. Distance has a correlation of 0.76 with the fare, rain 0.12 and rush hour about minus 0.08. The tip column has the highest value, 0.88. A correlation near 1 means the two columns rise together. The tip rises with the fare because it is a share of the fare, and it is only known after the trip, so using it to predict the fare would be cheating. That makes tip_ugx a leakage column.
