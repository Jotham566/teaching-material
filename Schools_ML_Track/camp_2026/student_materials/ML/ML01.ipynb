## Block B. Run a ready-made prediction

This model has looked at 120 invented homes and learned how rent follows floor area. You do not need to understand it yet. Session 13 opens it up.

**Predict.** Before you run it, write the monthly rent in UGX you expect for a home of 50 square metres. Write your guess in the cell below before you run anything.

My prediction:


```python
import pandas as pd
from sklearn.linear_model import LinearRegression
homes = pd.read_csv("synthetic_homes.csv")
model = LinearRegression().fit(homes[["area_m2"]], homes["rent_ugx"])
pred = model.predict(pd.DataFrame({"area_m2": [50]}))[0]
print("Estimated rent for 50 square metres:", round(pred))
```

**Run it, then compare.** The model estimates about UGX 309,260 a month. It drew a line through the 120 homes and read the rent at 50 square metres. Guessing a number from examples is called regression, and you will build one yourself in session 14. The homes are invented, so the estimate says nothing about real rents.
