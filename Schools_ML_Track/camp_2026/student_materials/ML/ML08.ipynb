## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| select a column | Name the column in square brackets to look only at it. | Reading only the price line of the register. |
| filter | Keep only the rows where a comparison is True. | Picking out the sales of 5 units or more. |
| new column | A column calculated from other columns, added to the table. | Multiplying units by price on every line. |
| sum | All the values added together. | Adding up the day's takings. |
| group | Gather rows with the same label before calculating. | Sorting the sales into three piles by item. |
| revenue | Money received from sales: units times price. It is not profit. | The takings before paying for stock. |

## Block B. A revenue column and totals by item

Multiply units by price on every row, then add the money for each item.

**Predict.** Predict which item earns the most money. Is it the one with the most units? Write your guess in the cell below before you run anything.

My prediction:


```python
shop["revenue_ugx"] = shop["units"] * shop["unit_price_ugx"]
total_revenue = shop["revenue_ugx"].sum()
by_item = shop.groupby("item")["revenue_ugx"].sum()
print(total_revenue)
print(by_item)
```

**Run it, then compare.** The total is 550000 shillings. Book earns 297000, Juice 150000 and Pen 103000. Pen sold the most units, 103, but Book earns the most money because each Book costs 3000.
