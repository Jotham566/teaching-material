## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| mean | The total shared out equally: add the values and divide by how many. | Sharing a bag of mandazi equally. |
| median | The middle value once the values are in order. | The pupil standing in the middle of a line sorted by height. |
| range | The largest value minus the smallest. | The gap between the tallest and the shortest. |
| percentile | The value below which a given share of the data falls. | The mark that 75 out of 100 pupils did not beat. |
| proportion | A part divided by the whole; times 100 it becomes a percentage. | 12 pupils out of 20 passed. |
| outlier | A value far from the rest. | One huge order in a week of small sales. |


## Block A. Typical value and spread

Look at how many units each sale has. Run the cell and read the four numbers.

**Predict.** Predict whether the mean or the median of the units column is larger. Write your guess in the cell below before you run anything.

My prediction:


```python
import numpy as np
import pandas as pd
shop = pd.read_csv("camp_shop.csv")
units = shop["units"]
mean_units = units.mean()
median_units = np.median(units)
spread = units.max() - units.min()
p75 = np.percentile(units, 75)
print(mean_units, median_units, spread, p75)
```

**Run it, then compare.** The cell prints about 4.62, 4.5, 7 and 7.0. The mean is the total of 277 divided by 60 sales. The median 4.5 sits between the 30th and 31st values in order. The range is 8 minus 1. The 75th percentile 7 means three quarters of the sales have 7 units or fewer. NumPy is the library behind pandas for numbers, and np is its short name.
