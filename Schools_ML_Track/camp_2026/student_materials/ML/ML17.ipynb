## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| overfitting | Learning the quirks of the training examples so well that new examples are predicted worse. | Memorising last year's exam instead of learning the subject. |
| confusion matrix | A table that counts actual categories against predicted categories. | A mark sheet of right and wrong answers by question type. |
| actual and predicted | Rows show what really happened; columns show what the model said. | The real result against the guess. |
| validation records | Held-out rows used to choose between models. | The practice test used to decide how to revise. |
| final test | The last check, run once after every choice is made. | The real exam. |
| locked test | A test set no choice has ever been based on. | An exam paper no student has seen. |

## Block B. The tree that remembered

Train trees that may ask more and more questions, and print their training and test scores.

**Predict.** Predict which tree scores 100% on the training flowers. Write your guess in the cell below before you run anything.

My prediction:


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.25, random_state=7)
for depth in [1, 2, 3, None]:
    model = DecisionTreeClassifier(max_depth=depth, random_state=7).fit(X_train, y_train)
    print(depth, round(model.score(X_train, y_train), 3), round(model.score(X_test, y_test), 3))
gap = model.score(X_train, y_train) - model.score(X_test, y_test)
```

**Run it, then compare.** Each line prints the depth, the training score and the test score. The tree with depth None may ask as many questions as it likes and scores 1.0 on the training flowers, because it memorised them. Its test score does not rise with it. That gap between training and test is the sign of overfitting. score gives the accuracy.
