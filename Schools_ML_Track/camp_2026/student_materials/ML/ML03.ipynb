## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| operator | A symbol that tells Python which calculation to do. | The plus and minus keys on a calculator. |
| parentheses | Brackets that make Python calculate the inside first. | Doing the part in brackets first in maths. |
| comment | A line starting with # that Python ignores and people read. | A note written in the margin. |
| convert | Change a value from one type to another, such as text to a number. | Changing shillings into dollars before adding. |
| error message | Python telling you what it could not do and where. | A teacher marking the exact line you got wrong. |
| traceback | The list of lines Python followed before the error. | A trail of footprints that ends at the problem. |
