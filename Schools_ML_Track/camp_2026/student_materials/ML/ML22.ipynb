## Block A. Run the final test once

The fares model with three inputs was chosen on validation. Now run the 30 locked test trips once.

**Predict.** Predict whether the final test error will be above or below the validation error of 2255. Write your guess in the cell below before you run anything.

My prediction:


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error
fares = pd.read_csv("synthetic_fares.csv")
X = fares[["distance_km", "rush_hour", "raining"]]
y = fares["fare_ugx"]
X_dev, X_test, y_dev, y_test = train_test_split(X, y, test_size=0.2, random_state=7)
X_train, X_val, y_train, y_val = train_test_split(X_dev, y_dev, test_size=0.25, random_state=7)
from sklearn.linear_model import LinearRegression
baseline = DummyRegressor(strategy="mean").fit(X_train, y_train)
model = LinearRegression().fit(X_train, y_train)
final_mae = mean_absolute_error(y_test, model.predict(X_test))
final_baseline_mae = mean_absolute_error(y_test, baseline.predict(X_test))
print(round(final_mae), round(final_baseline_mae))
```

**Run it, then compare.** The cell prints 2936 and 3617. On the locked 30 trips the model misses by about UGX 2936 and the plain guess by about UGX 3617. The model is still better, but its lead is much smaller than the validation lead of 2255 against 4361. This is normal: the test has only 30 trips, and the validation lead may have been lucky.
