## Block A. A function and an assert

A function names a calculation so you can use it again. Run the cell. A passing assert prints nothing.

**Predict.** Predict what total(2000, 3) gives and what happens when the assert passes. Write your guess in the cell below before you run anything.

My prediction:


```python
def total(price, quantity):
    return price * quantity

result = total(2000, 3)
print(result)
assert total(2000, 3) == 6000
```

## Block B. The flower rule as a function

Your rule from session 4 becomes a function. Three asserts test it, and a loop counts it on eight flowers.

**Predict.** Predict how many of the eight flowers the function names correctly. Write your guess in the cell below before you run anything.

My prediction:


```python
def guess_species(petal_length):
    if petal_length < 2.5:
        return "setosa"
    elif petal_length < 4.8:
        return "versicolor"
    else:
        return "virginica"

assert guess_species(1.4) == "setosa"
assert guess_species(4.7) == "versicolor"
assert guess_species(6.0) == "virginica"
flowers = [[1.4, "setosa"], [1.4, "setosa"], [4.7, "versicolor"], [4.5, "versicolor"], [4.9, "versicolor"], [6.0, "virginica"], [5.1, "virginica"], [5.9, "virginica"]]
right = 0
for flower in flowers:
    if guess_species(flower[0]) == flower[1]:
        right = right + 1
print(right)
```

**Run it, then compare.** The cell prints 7. The three asserts pass without a sound. Each flower in the list holds a petal length at position 0 and a species at position 1. The function is right for seven of eight flowers, the same count you found by hand.

1. What does flower[0] give?
2. Why test the function on three known flowers before counting?
