## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| comparison | A question about two values that answers True or False. | Asking whether your money is at least the ticket price. |
| Boolean | A value that is either True or False. | A light switch that is on or off. |
| if | Runs the indented lines below it only when its question is True. | If it rains, take an umbrella. |
| else | Runs when the if question was False. | Otherwise, leave the umbrella. |
| elif | Asks a second question when the first was False. | If it is raining take an umbrella, otherwise if it is cloudy take a hat. |
| indentation | Four spaces at the start of a line that show which instructions belong together. | Writing a sub-point under a heading. |

## Block B. A rule that names a flower

The iris data records flowers of three species. Petal length helps tell them apart. Run the rule for one flower, then try the others on the table.

**Predict.** Look at the rule. Predict which species it names for a petal length of 4.7 centimetres. Write your guess in the cell below before you run anything.

My prediction:


```python
petal_length = 4.7
if petal_length < 2.5:
    guess = "setosa"
elif petal_length < 4.8:
    guess = "versicolor"
else:
    guess = "virginica"
print(guess)
```
