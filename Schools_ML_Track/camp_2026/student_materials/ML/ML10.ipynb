|---|---|---|

## Block A. Count the problems

A tiny table has two sets of trouble. Count them before you change anything.

**Predict.** Predict how many missing prices and how many duplicate rows the raw table has. Write your guess in the cell below before you run anything.

My prediction:


```python
import pandas as pd
raw = pd.DataFrame({"id": [1, 2, 2, 3], "price": [1000, None, None, 3000]})
missing = raw["price"].isna().sum()
dupes = raw.duplicated().sum()
clean = raw.drop_duplicates().copy()
print(missing, dupes, len(clean))
print(clean["price"].mean())
```

**Run it, then compare.** The cell prints 2, 1 and 3, then 2000.0. isna marks each blank as True and sum counts them. duplicated marks the repeated row and drop_duplicates removes it. copy makes a separate table, so raw stays unchanged. The mean of 2000.0 uses the two prices that exist and ignores the missing one.
