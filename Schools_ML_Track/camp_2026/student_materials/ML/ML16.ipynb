## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| classification | Predicting a category instead of a number. | Deciding whether a fruit is a mango or an orange. |
| decision tree | A chain of yes or no questions about the features that ends in a category. | A flow chart that sorts the flowers. |
| class label | The name of a category; its code number is only an identifier. | The species name on a seed packet. |
| accuracy | The share of predictions that are right. | Marks scored out of the total. |
| majority baseline | A model that always names the most common category. | Always answering the most popular option. |
| depth | How many questions in a row the tree may ask. | How many forks a path may take. |


## Block A. Train a tree and draw it

Split the 150 iris flowers, train a tree that may ask two questions in a row, and draw it.

**Predict.** Predict whether the tree's accuracy will be above 85% and whether the majority baseline will be above 55%. Write your guess in the cell below before you run anything.

My prediction:


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.25, random_state=7)
tree = DecisionTreeClassifier(max_depth=2, random_state=7).fit(X_train, y_train)
baseline = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
tree_acc = accuracy_score(y_test, tree.predict(X_test))
base_acc = accuracy_score(y_test, baseline.predict(X_test))
print(round(tree_acc, 3), round(base_acc, 3))
plot_tree(tree, feature_names=iris.feature_names, class_names=list(iris.target_names))
plt.show()
```

**Run it, then compare.** The tree scores above 0.85 and the majority baseline scores about a third, because the three species are equally common. plot_tree draws the questions: the top box asks about a petal measurement, the same kind of question as your hand rule. The classes are coded 0, 1 and 2 for setosa, versicolor and virginica. The codes name categories, so species 2 is not twice species 1. Accuracy is correct predictions divided by all predictions.

## Block B. Back to the hand rule

Score your session 4 rule on all 150 flowers. Petal length is column 2 of the data. The rule says setosa below 2.5, versicolor below 4.8 and virginica otherwise.

**Predict.** Predict whether your hand rule will score above or below the tree. Write your guess in the cell below before you run anything.

My prediction:


```python
lengths = iris.data[:, 2]
guess = (lengths >= 2.5).astype(int) + (lengths >= 4.8).astype(int)
hand_acc = (guess == iris.target).mean()
print(round(hand_acc, 3))
```

**Run it, then compare.** The hand rule scores above 0.9 on all 150 flowers. Each comparison gives True or False for every flower and astype(int) turns them into 1 and 0. Adding the two gives 0 for setosa, 1 for versicolor and 2 for virginica. Your rule and the tree use the same idea. The tree chose its question and cut-off by itself, and you chose yours by looking at eight flowers.
