## Block A. Split the homes

Keep a quarter of the 120 homes back. The model must never see them while it learns.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
homes = pd.read_csv("synthetic_homes.csv")
X = homes[["area_m2", "rooms"]]
y = homes["rent_ugx"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=7)
print(len(X_train), len(X_test))
```

**Run it, then compare.** The cell prints 90 and 30. The four names on the left receive training inputs, test inputs, training targets and test targets, in that order. random_state=7 fixes the shuffle so everyone gets the same split.

## Block B. Baseline against model

Fit a mean-guess baseline and a line on the 90 training homes. Measure both on the 30 held-out homes.

**Predict.** Predict whether the model's average miss will be smaller than the baseline's, and by roughly how much. Write your guess in the cell below before you run anything.

My prediction:


```python
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
baseline = DummyRegressor(strategy="mean").fit(X_train, y_train)
model = LinearRegression().fit(X_train, y_train)
baseline_mae = mean_absolute_error(y_test, baseline.predict(X_test))
model_mae = mean_absolute_error(y_test, model.predict(X_test))
print(round(baseline_mae), round(model_mae))
```

**Run it, then compare.** The cell prints 110169 and 11889. The baseline misses by about UGX 110169 and the model by about UGX 11889, so on these 30 held-out homes the model is far better. Mean absolute error averages the misses and uses the unit of the target, shillings. The model wins by this much because the homes were generated from a formula, which makes them easy to learn. Real data would not be so kind.
