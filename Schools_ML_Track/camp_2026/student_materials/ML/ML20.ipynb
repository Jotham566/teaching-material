## Block A. Split and fit the baseline

Use the fares as the worked example, then repeat with your own dataset. Lock 20% away, take 25% of the rest for validation, and fit the mean guess on training only.

**Predict.** Predict the three sizes and whether the baseline will miss fares by more than UGX 3000 on average. Write your guess in the cell below before you run anything.

My prediction:


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error
fares = pd.read_csv("synthetic_fares.csv")
X = fares[["distance_km", "rush_hour", "raining"]]
y = fares["fare_ugx"]
X_dev, X_test, y_dev, y_test = train_test_split(X, y, test_size=0.2, random_state=7)
X_train, X_val, y_train, y_val = train_test_split(X_dev, y_dev, test_size=0.25, random_state=7)
baseline = DummyRegressor(strategy="mean").fit(X_train, y_train)
val_baseline_mae = mean_absolute_error(y_val, baseline.predict(X_val))
print(len(X_train), len(X_val), len(X_test), round(val_baseline_mae))
```

**Run it, then compare.** The cell prints 90, 30, 30 and 4361. The plain guess misses the validation fares by about UGX 4361 on average. Any model you build must beat that on the same validation trips. The baseline was fitted on the 90 training trips only. The 30 test trips stay locked.

## Block B. One descriptive chart

Plot the training trips: distance on one axis, fare on the other. Label both with units.

**Predict.** Predict the shape of the dots: rising, falling or no pattern. Write your guess in the cell below before you run anything.

My prediction:


```python
import seaborn as sns
import matplotlib.pyplot as plt
train = X_train.copy()
train["fare_ugx"] = y_train
sns.scatterplot(data=train, x="distance_km", y="fare_ugx")
plt.xlabel("Distance (km)")
plt.ylabel("Fare (UGX)")
plt.title("Fare against distance, training trips")
plt.show()
train_corr = train["distance_km"].corr(train["fare_ugx"])
```

**Run it, then compare.** The dots rise from left to right with a lot of scatter. The correlation on the 90 training trips is about 0.76. Longer trips cost more, but two trips of equal length can differ by thousands of shillings. That scatter is why even the best line will still miss.
