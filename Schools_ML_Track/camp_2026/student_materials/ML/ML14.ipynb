# ML14 Predicting a number with a line

**Today's question.** How close does a straight line get to the rents of eight homes?

**Objective.** By the end: you can guess a line by hand, measure its error, and compare it with the line scikit-learn fits.

## Words for today

| Word | What it means | Think of it like this |
|---|---|---|
| regression | Predicting a number. | Estimating a price. |
| slope | How much the prediction changes when the input rises by one unit. | Extra rent for every extra square metre. |
| intercept | The prediction when the input is zero. | The starting fee before any distance is added. |
| fit | Let the computer find the best slope and intercept from examples. | Practising until the throws land close. |
| predict | Use the fitted line on a new input. | Reading the answer off the line. |
| error | The gap between a prediction and the real value. | How far off your estimate was. |


## Block A. Guess a line, measure the error

Take the first eight homes. Guess that rent is 100000 plus 3000 for every square metre.

**Predict.** Predict whether the average error will be under UGX 10000 or over UGX 50000. Write your guess in the cell below before you run anything.

My prediction:


```python
import pandas as pd
homes = pd.read_csv("synthetic_homes.csv").head(8)
guess = 100000 + 3000 * homes["area_m2"]
errors = (homes["rent_ugx"] - guess).abs()
print(errors.mean())
```

**Run it, then compare.** The cell prints 56282.25. Your guess is a straight line with intercept 100000 and slope 3000 shillings per square metre. On average it misses the real rent by about UGX 56282. The line ignores rooms, and the intercept is too low, so most guesses are too small.

## Block B. Let scikit-learn fit the line

Ask LinearRegression to find the best slope and intercept for the same eight homes. Compare its error with the baseline.

**Predict.** Predict whether the fitted line's error is smaller or larger than your hand-guessed 56282. Write your guess in the cell below before you run anything.

My prediction:


```python
from sklearn.linear_model import LinearRegression
X = homes[["area_m2"]]
y = homes["rent_ugx"]
model = LinearRegression().fit(X, y)
fit_error = (y - model.predict(X)).abs().mean()
baseline_error = (y - y.mean()).abs().mean()
print(round(model.coef_[0]), round(model.intercept_), round(fit_error), round(baseline_error))
```

**Run it, then compare.** The cell prints 3221, 135838, 20077 and 109399. The fitted line has slope 3221 and intercept 135838 and misses by about UGX 20077 on these eight homes. Guessing the mean every time would miss by about UGX 109399. fit found the line, and predict will use it on a new input. These errors are measured on the eight homes the line learned from, so they look better than they will on new homes.

2. How much better is the fitted line than the baseline?
