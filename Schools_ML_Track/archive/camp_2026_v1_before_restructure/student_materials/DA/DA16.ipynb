## Learning objective

By the end of this class: Join on a key, check row counts and explain how duplicate keys could inflate revenue.

## Understand the idea

A key identifies a row. A join connects records using matching keys. sales.item matches products.item here; products has exactly one row per item. LEFT JOIN keeps every sale even if product details are missing. A duplicate product key would multiply matching sale rows and overstate totals.



## Worked task
SELECT s.sale_id, s.item, p.category FROM sales AS s LEFT JOIN products AS p ON s.item = p.item ORDER BY s.sale_id;

Expected result: 60 joined rows, one per sale, with a category from the product table.

## Your task

Repair cue: Check unique keys before joining. If the count increases above 60, investigate duplicate product keys instead of accepting the larger total.
