## Understand the idea

NULL means missing or unknown in a database. Test it with IS NULL, not = NULL. Validation checks whether results obey expectations: row counts, allowed values and reconciled totals. Saving a query preserves the calculation; saving only a result does not preserve its method.



## Worked task
SELECT COUNT(*) AS missing_units FROM sales WHERE units IS NULL;
SELECT MIN(units) AS smallest, MAX(units) AS largest FROM sales;

Expected result: No missing units in the supplied clean practice file. Minimum and maximum can be checked against the raw CSV.
